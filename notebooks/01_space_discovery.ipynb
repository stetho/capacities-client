# 01 — Space Discovery

First contact with the real API. The goal here is simply to see what our Capacities space actually looks like — what structures exist, what properties they have, and whether our client library works against live data.

```python
import sys
sys.path.insert(0, '..')

from capacities.client import CapacitiesClient
from capacities.space import SpaceAPI
```

```python
# Token loads from .env automatically
client = CapacitiesClient()
space_api = SpaceAPI(client)

spaces = space_api.get_spaces()
for s in spaces:
    print(f"{s.id}  {s.title}")
```

```python
# Paste your space ID from the output above
SPACE_ID = "a48d0ff0-6826-4fab-8e14-83ea02c97586"

space = space_api.get_space_info(SPACE_ID)
print(f"Space: {space.title}\n")
for structure in space.structures:
    print(f"  {structure.title} ({structure.id})")
    for prop in structure.properties:
        print(f"    · {prop.name} [{prop.type}]")
```

```python
# Keep the raw response too — useful for spotting anything our models missed
import json

raw = client._get("/space-info", params={"spaceid": SPACE_ID})
print(json.dumps(raw, indent=2))
```
